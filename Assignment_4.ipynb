```python
import pandas as pd
import numpy as np

data=pd.DataFrame({
    'Color':['Red','Blue','Red','Green'],
    'City':['Delhi','Mumbai','Delhi','Kolkata']
})
print("Nominal Dataset")
print(data)

n=len(data)
matrix=np.zeros((n,n))

for i in range(n):
  for j in range(n):
    count=0

    for col in data.columns:
      if data.loc[i,col]!=data.loc[j,col]:
        count+=1

      matrix[i,j]=count/len(data.columns)
print("\nDissimilarity matrix - Nominal")
print(pd.DataFrame(
    matrix,
    index=['01','02','03','04'],
    columns=['01','02','03','04']
    ).round(2))
```

```python
import pandas as pd
import numpy as np

data=pd.DataFrame({
    'Student':[1,1,0,0],
    'Employed':[0,1,1,0],
    'Purchased':[1,1,0,1]
 })

print("Binary Dataset")
print(data)

n=len(data)
matrix=np.zeros((n,n))

for i in range(n):
  for j in range(n):
    diff=0

    for col in data.columns:
      if data.loc[i,col]!=data.loc[j,col]:
        diff+=1

      matrix[i,j]=diff/len(data.columns)

print("\nDissimilarity Matrix - Binary")
print(pd.DataFrame(
    matrix,
    index=['01','02','03','04'],
    columns=['01','02','03','04']
    ).round(2))
```

```python
import pandas as pd
import numpy as np

data=pd.DataFrame({
    'Satisfaction':['Low','Medium','High','Medium']
 })

print("Ordinal Dataset")
print(data)

rank={
    'Low':0,
    'Medium':1,
    "High":2
}

n=len(data)
matrix=np.zeros((n,n))

for i in range(n):
  for j in range(n):
    x=rank[data.loc[i,'Satisfaction']]
    y=rank[data.loc[j,'Satisfaction']]
    matrix[i,j]=abs(x-y)/2

print("\nDissimilarity Matrix - Ordinal")
print(pd.DataFrame(
    matrix,
    index=['01','02','03','04'],
    columns=['01','02','03','04']
    ).round(2))
```
